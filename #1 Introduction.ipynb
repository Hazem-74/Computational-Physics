### Problem 3: Quadratic Function Plot

$$ f(x) = x^2 − 3x + 2, \quad -2 ≤ x ≤ 4 $$

```python
# Write a function to compute
x = np.linspace (-2 , 4 , 100)
y = x**2 - 3*x + 2

# roots
roots = np.array([1, 2])
y_roots = np.zeros_like(roots)

# plot
plt.plot(x,y,label=r"$ f(x) = x^2 − 3x + 2 $")
plt.scatter(roots, y_roots, color='red', label="Roots")
plt.xlabel("x")
plt.ylabel(r"$f(x)$")
plt.title(r"$f(x) = x^2 − 3x + 2$")
plt.legend()
plt.grid(True)
plt.show()
```

### Problem 5: Monte Carlo Approximation of $\pi$

```python
# Use random number generation to estimate the value of π 
# by simulating points in a square 
# and counting how many fall inside the unit circle

n = 100000
x = np.random.uniform(-1, 1, n)
y = np.random.uniform(-1, 1, n)

unit_circle = (x**2 + y**2)

unit_circle = (unit_circle <= 1)

pi  = 4 * np.sum(unit_circle) / n

print(f"π ≈ {pi}")
```

```python
# Plot the scatter diagram of inside vs. outside points 
# Ref[1] was used

plt.scatter(x[unit_circle], y[unit_circle], color='blue', label='Inside Circle')
plt.scatter(x[~unit_circle], y[~unit_circle], color='red', label='Outside Circle')

# Draw the circle outline for reference
circle = plt.Circle((0, 0), 1, color='green', fill=False, linewidth=2)
plt.gca().add_artist(circle)

plt.xlabel("x")
plt.ylabel("y")
plt.title(f"Monte Carlo Estimation of π ≈ {pi}")
plt.legend()
plt.show()
```

### Problem 6: Function Exploration

$$ f(t) = e^{−0.1t} sin(2πt), \quad 0 ≤ t ≤ 10 $$

```python
# Write Python code to compute and plot the function

t = np.linspace(0, 10, 100)
y = np.exp(-0.1 * t) * np.sin(2 * np.pi * t)

# plot

plt.plot(t, y, label=r"$ f(t) = e^{−0.1t} sin(2πt) $")
plt.xlabel("t")
plt.ylabel(r"f(t)")
plt.title(r"$ f(t) = e^{−0.1t} sin(2πt) $")
plt.legend()
plt.grid(True)
plt.show()
```

#### Physical interpretation:

- The coefficient *0.1* in the exponent represents **the damping constant** which controls how fast the amplitude decay and the rate of energy loss in oscillating systems.

```python
# Another comparison (reversing the factors in functions)
y_sin = np.exp(-c2 * t) * np.sin(2 * np.pi * t)
y_cos = np.exp(-c1 * t) * np.cos(2 * np.pi * t)

# Plot
plt.plot(t, y_sin, label=rf"$e^{{-{c2}t}}\sin(2\pi t)$", color="blue")
plt.plot(t, y_cos, label=rf"$e^{{-{c1}t}}\cos(2\pi t)$", color="red")
plt.xlabel("t")
plt.ylabel(r"f(t)")
plt.title("Comparison of Damped Sine and Cosine Oscillations (revered factors)")
plt.legend()
plt.grid(True)
plt.show()
```

#### Physical interpretation:

- Damping affects both amplitude and duration, regardless it’s sine or cosine but the starting phase determines how the decay begins.

- **The cosine wave** *(starts at maximum amplitude)* shows the immediate effect of damping, so its peak visibly shrinks faster when damping is high.

- **The sine wave** *(starts at zero)* initially rises before damping dominates, so the decay appears slightly delayed visually.
